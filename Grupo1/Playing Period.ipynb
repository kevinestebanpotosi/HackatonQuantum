## Challenge Statement

The Quantum Fourier Transform (QFT) is one of the most important subroutines in quantum computing. It is used in the most famous algorithms such as quantum phase estimation and Shor's factoring algorithm!

For a system of $n$ qubits, the QFT transforms a computational basis state $\vert{}x\rangle$ according to:

$$\vert{}x\rangle \longrightarrow \frac{1}{\sqrt{2^n}} \sum_{y=0}^{2^n-1} e^{2\pi i x y / 2^n} \vert{}y\rangle.$$

This transformation exists in PennyLane, but in this challenge you are going to build it from elementary gates, namely Hadamard gates (`qp.Hadamard`), controlled phase rotations (`qp.ControlledPhaseShift`), and SWAP gates (`qp.SWAP`). We will focus on the four-qubit QFT.
